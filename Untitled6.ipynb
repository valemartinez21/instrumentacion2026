```python
import numpy as np
import matplotlib.pyplot as plt

from astropy.io import fits
from astropy.wcs import WCS
```

```python
# Abrir imágenes FITS

star_1b = fits.open("Star1_B.fits")[0]
star_1v = fits.open("Star1_V.fits")[0]
star_2b = fits.open("Star2_B.fits")[0]
star_2v = fits.open("Star2_V.fits")[0]

# Crear sus transformaciones WCS
wcs_1b = WCS(star_1b.header).celestial
wcs_1v = WCS(star_1v.header).celestial
wcs_2b = WCS(star_2b.header).celestial
wcs_2v = WCS(star_2v.header).celestial

print("WCS cargadas correctamente")
```

```python
# Mostrar imagen con coordenadas celestes

fig = plt.figure(figsize=(9, 8))

ax = fig.add_subplot(
    111,
    projection=wcs_1v
)

ax.imshow(
    star_1v.data,
    origin="lower",
    cmap="gray",
    vmin=np.nanpercentile(star_1v.data, 5),
    vmax=np.nanpercentile(star_1v.data, 99.5)
)

ax.coords[0].set_axislabel("Ascensión recta")
ax.coords[1].set_axislabel("Declinación")

ax.coords.grid(color="cyan", alpha=0.5)

plt.title("Astrometría inicial - HIP 6257 V")
plt.show()
```

```python

from astropy.io import fits
from astropy.wcs import WCS
import matplotlib.pyplot as plt

# Archivo resuelto por Astrometry.net
archivo = "Img_B_resuelta.fits"

with fits.open(archivo) as hdul:
    imagen = hdul[0].data
    header = hdul[0].header.copy()

# Construir WCS
wcs = WCS(header).celestial

# Coordenadas del centro
ny, nx = imagen.shape
ra, dec = wcs.pixel_to_world_values(
    (nx - 1) / 2, (ny - 1) / 2
)

print(f"Ascensión recta: {ra:.6f} grados")
print(f"Declinación: {dec:.6f} grados")

# Visualización con coordenadas celestes
fig = plt.figure(figsize=(9, 8))
ax = fig.add_subplot(111, projection=wcs)

ax.imshow(imagen, origin="lower",
          cmap="gray",
          vmin=0,
          vmax=500)

ax.set_xlabel("Ascensión recta")
ax.set_ylabel("Declinación")
ax.set_title("NGC 253 - Calibración astrométrica")
ax.grid(color="white", alpha=0.4)

plt.show()
```
